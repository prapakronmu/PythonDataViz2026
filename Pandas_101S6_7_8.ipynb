### สามวิธีในการ `import` ใน Python

การ `import` ใน Python ใช้เพื่อนำโมดูล (module) หรือส่วนประกอบต่างๆ จากไฟล์ Python อื่นๆ มาใช้งานในโค้ดปัจจุบันของเรา เพื่อให้โค้ดของเราเป็นระเบียบ, สามารถนำกลับมาใช้ใหม่ได้ (reusable) และไม่ซ้ำซ้อนกันค่ะ มี 3 วิธีหลักๆ ดังนี้:

1.  **`import module_name`**: **นำเข้าทั้งโมดูล**
    
    *   **วัตถุประสงค์**: ใช้นำเข้าทั้งโมดูลเข้ามาในโปรเจกต์ของคุณ เมื่อคุณต้องการใช้ฟังก์ชัน, คลาส, หรือตัวแปรใดๆ ที่อยู่ในโมดูลนั้น คุณจะต้องระบุชื่อโมดูลนำหน้าเสมอ
    *   **ข้อดี**: ป้องกันชื่อซ้ำซ้อน (namespace collision) เพราะคุณรู้เสมอว่าอะไรมาจากโมดูลไหน
    *   **ข้อเสีย**: อาจจะต้องพิมพ์ชื่อโมดูลยาวๆ หลายครั้งหากมีการใช้งานบ่อย
    *   **ตัวอย่าง**: หากต้องการใช้ `DataFrame` จากโมดูล `pandas`
        ```python
        import pandas
        df = pandas.DataFrame() # ต้องเรียกใช้ผ่านชื่อโมดูล pandas
        ```
        ในโค้ดของคุณ: `import pandas # คือธรรมดา` (cell `K7EMFzOz7PJd`)

2.  **`import module_name as alias`**: **นำเข้าทั้งโมดูลและตั้งชื่อย่อ**
    
    *   **วัตถุประสงค์**: เป็นวิธีที่ได้รับความนิยมมาก โดยเฉพาะกับไลบรารียอดนิยมอย่าง `pandas` (มักจะตั้งชื่อย่อว่า `pd`) หรือ `numpy` (มักจะตั้งชื่อย่อว่า `np`) เพื่อให้โค้ดสั้นลง อ่านง่ายขึ้น และยังคงประโยชน์ในการป้องกันชื่อซ้ำซ้อนอยู่
    *   **ข้อดี**: โค้ดสั้นลง, อ่านง่ายขึ้น และยังคงชัดเจนว่าฟังก์ชันนั้นๆ มาจากโมดูลใด
    *   **ตัวอย่าง**: หากต้องการใช้ `DataFrame` จากโมดูล `pandas` โดยใช้ชื่อย่อว่า `pd`
        ```python
        import pandas as pd
        df = pd.DataFrame() # เรียกใช้ผ่านชื่อย่อ pd
        ```
        ในโค้ดของคุณ: `import pandas as pd #แบบย่อ` (cell `-db5CJRNlwyL`)

3.  **`from module_name import item_name`** (หรือ **`from module_name import item_1, item_2`** หรือ **`from module_name import *`**): **นำเข้าเฉพาะบางส่วนของโมดูล**
    
    *   **วัตถุประสงค์**: ใช้นำเข้าเฉพาะฟังก์ชัน, คลาส, หรือตัวแปรที่คุณต้องการใช้จากโมดูลนั้นๆ โดยตรง เข้ามาในเนมสเปซปัจจุบัน (current namespace) ของโค้ดคุณ คุณไม่จำเป็นต้องระบุชื่อโมดูลนำหน้าเมื่อเรียกใช้งาน
    *   **รูปแบบต่างๆ**:
        *   **`from module_name import item_name`**: นำเข้าเพียง 1 ส่วนที่ต้องการ
            ```python
            from pandas import DataFrame
            df = DataFrame() # เรียกใช้ได้โดยตรง ไม่ต้องมี pandas. นำหน้า
            ```
        *   **`from module_name import item_1, item_2`**: นำเข้าหลายส่วนที่ต้องการพร้อมกัน
            ```python
            from os import path, listdir
            print(path.join('a', 'b')) # ไม่ต้อง os.path.join
            ```
        *   **`from module_name import *`**: **นำเข้าทุกอย่าง** นี่คือการนำทุกอย่างในโมดูลนั้นเข้ามาในเนมสเปซปัจจุบันของคุณ **โดยทั่วไปแล้วไม่แนะนำให้ใช้** โดยเฉพาะในโปรเจกต์ขนาดใหญ่ เพราะอาจทำให้เกิดปัญหาชื่อซ้ำซ้อนได้ยากที่จะแก้ไข (เช่น คุณอาจมีฟังก์ชันชื่อ `foo()` ในโค้ดของคุณ และโมดูลที่คุณนำเข้าก็มีฟังก์ชันชื่อ `foo()` เช่นกัน ซึ่งจะทำให้เกิดความสับสนว่ากำลังเรียกใช้ฟังก์ชันใด)

**สรุปความแตกต่างและเวลาที่เหมาะสมในการใช้แต่ละแบบ:**

*   **`import module_name`** หรือ **`import module_name as alias`**: เหมาะสำหรับโมดูลที่คุณจะใช้งานบ่อย และต้องการความชัดเจนว่าฟังก์ชันหรือคลาสมาจากโมดูลใด เพื่อหลีกเลี่ยงความสับสนและชื่อซ้ำซ้อน

## Quiz ตัดตารางออกมาเฉพาะปี 2021 ```announce_date``` ในปี 2021

**Hint**
*   วนลูปหา index ของปี 2021
*   ตัดตารางมาเฉพาะ ปี 2021



````
# สร้าง List เปล่ามารอเก็บค่า True/False
TF = list()

# วนลูปเช็คทีละแถวในตาราง data_covid
for each_row in data_covid.iterrows():
    # แยกวันที่ด้วยเครื่องหมาย '/' แล้วดึงตำแหน่งปี (index ที่ 2) มาเทียบว่าเป็น '2021' หรือไม่
    if each_row[1]['announce_date'].split('/')[2] == '2021':
        TF.append(True)
    else:
        TF.append(False)

# ตัดตารางเอาเฉพาะแถวที่เป็น True (คือปี 2021)
data_covid_2021 = data_covid[TF]

# ดูผลลัพธ์ 5 แถวแรก
data_covid_2021.head()
````
